Your shell and notebook kernel can have different working directories. Run this cell inside the notebook to inspect its environment.

```python
from pathlib import Path
from tempfile import TemporaryDirectory
import sys

print("Notebook folder:", Path.cwd().name)
print(f"Notebook Python: {sys.version_info.major}.{sys.version_info.minor}")
print("Virtual environment active:", sys.prefix != sys.base_prefix)
```

Use a temporary directory to repeat the file exercise in Python. Convert each text row to an integer before summing the counts.

```python
with TemporaryDirectory() as temporary:
    path = Path(temporary) / "multiplicity.txt"
    path.write_text("0\n2\n5\n", encoding="utf-8")
    counts = [int(row) for row in path.read_text(encoding="utf-8").splitlines()]

print(f"{len(counts)} events, {sum(counts)} charged particles")
```
