# Plot particle spectra

In 40 minutes, make three histograms for final-state charged particles. Transverse momentum $p_T=\sqrt{p_x^2+p_y^2}$ measures momentum perpendicular to the beam. Pseudorapidity $\eta=-\ln\tan(\theta/2)$ describes direction relative to the positive beam axis. Large $|\eta|$ is close to the beam.

Select $|\eta|<2.5$ and $p_T>0.2$ GeV. These are generator-level cuts. The sample has no detector simulation.

```python
import matplotlib.pyplot as plt
from hep_lab.simulation import generate

events = generate("soft", n_events=500, seed=42)
selected = [
    [p for p in event if p.charge != 0 and p.pt > 0.2 and abs(p.eta) < 2.5]
    for event in events
]
multiplicity = [len(event) for event in selected]
pt = [p.pt for event in selected for p in event]
eta = [p.eta for event in selected for p in event]

assert sum(multiplicity) == len(pt) == len(eta)
print(f"{len(events)} events, {len(pt)} selected particles, {multiplicity.count(0)} empty selections")
```

The $p_T$ and $\eta$ plots get one entry per selected particle. The multiplicity plot gets one entry per event, including zero. Report the count above 5 GeV, outside the $p_T$ panel.

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 3.5), layout="constrained")
axes[0].hist(pt, bins=25, range=(0, 5), color="#007580")
axes[0].set(xlabel=r"$p_T$ [GeV]", ylabel="Particles / bin")
axes[1].hist(eta, bins=25, range=(-2.5, 2.5), color="#007580")
axes[1].set(xlabel=r"$\eta$", ylabel="Particles / bin")
axes[2].hist(multiplicity, bins=[n - 0.5 for n in range(max(multiplicity) + 2)], color="#007580")
axes[2].set(xlabel="Charged multiplicity", ylabel="Events / bin")
print(f"Above pT plot range: {sum(value > 5 for value in pt)} particles")
plt.show()
```
