# Generate proton–proton collisions with Pythia 8

In 35 minutes, simulate 100 proton–proton collisions at a centre-of-mass energy of 13 TeV. Monte Carlo generation samples events from a physics model. Reusing a seed with the same generator build and settings lets you repeat a run.

Read `src/hep_lab/simulation.py`. It configures the beams, enables soft non-diffractive QCD collisions and calls `init()` once. Every `next()` attempts to generate a new event. The full event record includes incoming beams, intermediate states and final particles. The helper selects with `isFinal()` and copies the particle values before the next event replaces the record.

[PYTHIA 8.3 documentation](https://pythia.org/)

```python
from collections import Counter
from hep_lab.simulation import generate

events = generate("soft", n_events=100, seed=42)
assert len(events) == 100
print(f"{len(events)} events. First event has {len(events[0])} final particles.")
for particle in events[0][:8]:
    print(f"PDG ID {particle.pdg_id:5d}  charge {particle.charge:+.0f}  pT {particle.pt:.2f} GeV")
```

A PDG ID identifies a species. For example 211 is a positive pion, −211 a negative pion and 22 a photon. The sign of a PDG ID follows a particle–antiparticle convention. Use the charge field to select charged particles.

From Terminal, repeat the run with `python -m hep_lab.simulation --events 100 --seed 42`.

```python
ids = Counter(p.pdg_id for event in events for p in event)
print("Five most frequent final-state PDG IDs:", ids.most_common(5))
```
